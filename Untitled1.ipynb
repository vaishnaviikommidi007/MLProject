```python
# Sort data chronologically for each Store-SKU combination

df = df.sort_values(
    ["store_id", "sku_id", "date"]
).reset_index(drop=True)

print("Data sorted successfully!")
print(df[["store_id", "sku_id", "date", "units_sold"]].head(10))
```

```python
# Create lag features for each Store-SKU combination

group = df.groupby(["store_id", "sku_id"])["units_sold"]

df["Lag_1"] = group.shift(1)
df["Lag_7"] = group.shift(7)
df["Lag_14"] = group.shift(14)
df["Lag_28"] = group.shift(28)

print(df[[
    "store_id",
    "sku_id",
    "date",
    "units_sold",
    "Lag_1",
    "Lag_7",
    "Lag_14",
    "Lag_28"
]].head(35))
```

```python
# Create rolling demand features

group = df.groupby(["store_id", "sku_id"])["units_sold"]

df["Rolling_Mean_7"] = (
    group.transform(lambda x: x.shift(1).rolling(7).mean())
)

df["Rolling_Mean_14"] = (
    group.transform(lambda x: x.shift(1).rolling(14).mean())
)

df["Rolling_Mean_28"] = (
    group.transform(lambda x: x.shift(1).rolling(28).mean())
)

df["Rolling_Std_7"] = (
    group.transform(lambda x: x.shift(1).rolling(7).std())
)

print(df[[
    "date",
    "store_id",
    "sku_id",
    "units_sold",
    "Lag_1",
    "Lag_7",
    "Rolling_Mean_7",
    "Rolling_Mean_14",
    "Rolling_Mean_28",
    "Rolling_Std_7"
]].head(35))
```

```python
print("NaN values after creating lag and rolling features:")

print(
    df[[
        "Lag_1",
        "Lag_7",
        "Lag_14",
        "Lag_28",
        "Rolling_Mean_7",
        "Rolling_Mean_14",
        "Rolling_Mean_28",
        "Rolling_Std_7"
    ]].isnull().sum()
)
```

```python
# Create price and promotion features

df["discount_amount"] = (
    df["list_price"] * df["discount_pct"] / 100
)

df["selling_price"] = (
    df["list_price"] - df["discount_amount"]
)

df["promo_discount"] = (
    df["discount_pct"] * df["promo_flag"]
)

print(df[[
    "list_price",
    "discount_pct",
    "promo_flag",
    "discount_amount",
    "selling_price",
    "promo_discount"
]].head(10))
```

```python
print("Price and promotion feature statistics:")

print(df[[
    "list_price",
    "discount_pct",
    "discount_amount",
    "selling_price",
    "promo_discount"
]].describe().T)
```

```python
# Create inventory-related features

df["low_stock_flag"] = (
    df["stock_on_hand"] <= 20
).astype(int)

df["inventory_value"] = (
    df["stock_on_hand"] * df["purchase_cost"]
)

df["stockout_promo"] = (
    df["stock_out_flag"] * df["promo_flag"]
)

print(df[[
    "stock_on_hand",
    "stock_out_flag",
    "lead_time_days",
    "purchase_cost",
    "low_stock_flag",
    "inventory_value",
    "stockout_promo"
]].head(10))
```

```python
print("Inventory feature statistics:")

print(df[[
    "stock_on_hand",
    "lead_time_days",
    "purchase_cost",
    "low_stock_flag",
    "inventory_value",
    "stockout_promo"
]].describe().T)
```

```python
print("Low stock rows:", df["low_stock_flag"].sum())
print("Stockout rows:", df["stock_out_flag"].sum())
print("Stockout + promotion rows:", df["stockout_promo"].sum())
```

```python
# Inventory coverage features

df["inventory_cover_7"] = (
    df["stock_on_hand"] / (df["Rolling_Mean_7"] + 1e-6)
)

df["inventory_cover_28"] = (
    df["stock_on_hand"] / (df["Rolling_Mean_28"] + 1e-6)
)

df["low_inventory_cover_flag"] = (
    df["inventory_cover_7"] < 3
).astype(int)

print(df[[
    "stock_on_hand",
    "Rolling_Mean_7",
    "Rolling_Mean_28",
    "inventory_cover_7",
    "inventory_cover_28",
    "low_inventory_cover_flag"
]].head(15))
```

```python
required_history_features = [
    "Lag_1",
    "Lag_7",
    "Lag_14",
    "Lag_28",
    "Rolling_Mean_7",
    "Rolling_Mean_14",
    "Rolling_Mean_28",
    "Rolling_Std_7",
    "inventory_cover_7",
    "inventory_cover_28"
]

valid_rows = df[required_history_features].notna().all(axis=1)

print("Total rows:", len(df))
print("Rows with complete history:", valid_rows.sum())
print("Rows removed:", (~valid_rows).sum())
print(
    "Percentage retained:",
    valid_rows.mean() * 100
)
```

```python
# Keep only rows with complete historical features

df_model = df.loc[valid_rows].copy()

print("Final modeling dataset shape:")
print(df_model.shape)

print("\nRemaining NaNs:")
print(df_model.isna().sum().sum())
```

```python
print("Modeling date range:")
print("Start:", df_model["date"].min())
print("End:", df_model["date"].max())
```

```python
print("\nTarget statistics:")
print(df_model["units_sold"].describe())
```

```python
features = [
    # Time
    "year",
    "month",
    "day",
    "weekofyear",
    "weekday",
    "is_weekend",
    "is_holiday",

    # Product / location
    "store_id",
    "sku_id",
    "category",
    "subcategory",
    "brand",
    "channel",

    # Weather
    "temperature",
    "rain_mm",

    # Price / promotion
    "list_price",
    "discount_pct",
    "selling_price",
    "promo_flag",
    "promo_discount",

    # Inventory
    "stock_on_hand",
    "stock_out_flag",
    "lead_time_days",
    "low_stock_flag",
    "inventory_cover_7",
    "inventory_cover_28",
    "low_inventory_cover_flag",

    # Historical demand
    "Lag_1",
    "Lag_7",
    "Lag_14",
    "Lag_28",
    "Rolling_Mean_7",
    "Rolling_Mean_14",
    "Rolling_Mean_28",
    "Rolling_Std_7"
]

target = "units_sold"

print("Number of features:", len(features))
```

```python
print("Final modeling dataset shape:")
print(df_model.shape)

print("\nRemaining NaNs:")
print(df_model[features].isna().sum().sum())

print("\nDate range:")
print("Start:", df_model["date"].min())
print("End:", df_model["date"].max())
```

```python
print("\nRows by year:")
print(df_model.groupby("year").size())
```

```python
# Chronological 80/20 train-test split

unique_dates = np.sort(df_model["date"].unique())

split_index = int(len(unique_dates) * 0.80)

cutoff_date = unique_dates[split_index]

train_df = df_model[
    df_model["date"] < cutoff_date
].copy()

test_df = df_model[
    df_model["date"] >= cutoff_date
].copy()

print("Cutoff date:", cutoff_date)

print("\nTrain shape:", train_df.shape)
print("Test shape:", test_df.shape)

print("\nTrain date range:")
print(train_df["date"].min(), "to", train_df["date"].max())

print("\nTest date range:")
print(test_df["date"].min(), "to", test_df["date"].max())
```

```python
total_rows = len(train_df) + len(test_df)

print("\nTrain percentage:",
      len(train_df) / total_rows * 100)

print("Test percentage:",
      len(test_df) / total_rows * 100)
```

```python
# Create training and testing features/target

X_train = train_df[features]
y_train = train_df["units_sold"]

X_test = test_df[features]
y_test = test_df["units_sold"]

print("X_train shape:", X_train.shape)
print("y_train shape:", y_train.shape)

print("X_test shape:", X_test.shape)
print("y_test shape:", y_test.shape)
```

```python
categorical_features = [
    "store_id",
    "sku_id",
    "category",
    "subcategory",
    "brand",
    "channel"
]

numerical_features = [
    col for col in features
    if col not in categorical_features
]

print("Categorical features:")
print(categorical_features)

print("\nNumber of categorical features:",
      len(categorical_features))

print("\nNumber of numerical features:",
      len(numerical_features))
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder

preprocessor = ColumnTransformer(
    transformers=[
        (
            "cat",
            OneHotEncoder(
                handle_unknown="ignore",
                sparse_output=True
            ),
            categorical_features
        ),
        (
            "num",
            "passthrough",
            numerical_features
        )
    ]
)

X_train_encoded = preprocessor.fit_transform(X_train)
X_test_encoded = preprocessor.transform(X_test)

print("Encoded X_train shape:", X_train_encoded.shape)
print("Encoded X_test shape:", X_test_encoded.shape)
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler(with_mean=False)

X_train_scaled = scaler.fit_transform(X_train_encoded)
X_test_scaled = scaler.transform(X_test_encoded)

print("Scaled X_train shape:", X_train_scaled.shape)
print("Scaled X_test shape:", X_test_scaled.shape)
```

```python
StandardScaler(with_mean=False)
```

```python
from sklearn.linear_model import LinearRegression

lr_model = LinearRegression()

print("Training Linear Regression...")

lr_model.fit(
    X_train_scaled,
    y_train
)

print("Training completed!")
```

```python
lr_pred = lr_model.predict(X_test_scaled)

print("Predictions generated!")
print("Number of predictions:", len(lr_pred))
```

```python
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

lr_mae = mean_absolute_error(y_test, lr_pred)

lr_rmse = np.sqrt(
    mean_squared_error(y_test, lr_pred)
)

lr_r2 = r2_score(y_test, lr_pred)

lr_mape = np.mean(
    np.abs(
        (y_test - lr_pred) /
        np.where(y_test == 0, 1, y_test)
    )
) * 100

print("Linear Regression Results")
print("-------------------------")
print("MAE :", lr_mae)
print("RMSE:", lr_rmse)
print("R²  :", lr_r2)
print("MAPE:", lr_mape, "%")
```
